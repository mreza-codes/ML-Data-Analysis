```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy import nan as NA

x = pd.read_csv('Titanic-Dataset.csv')  #get the dataset and place it near the notebook
print('x = \n',x.columns)
print()
print(x.dtypes)
print()
print(x.shape)
mean_age = np.mean(x['Age'])
print('average age = \n',mean_age)
print(x['Sex'].unique())
counter = 0
print(x['Survived'].head(10))
for i in x['Survived'] :
    if i == 1 :
        counter += 1
       
print('counter = ',counter)
print()
print(x['Sex'].unique())
print()

female_survive = len (x[(x['Sex'] == 'female') & (x['Survived'] == 1)])
print('number of female that survived = \n',female_survive)
print()
male_survive = len (x[(x['Sex'] == 'male')&(x['Survived'] == 1)])
print('number of men that survived = \n',male_survive)
```

Survival Percentages by Gender — Basic Calculations

```python
print('x = \n',x.columns)
print()

female_survive = len (x[(x['Sex'] == 'female') & (x['Survived'] == 1)])
print(' mean number of female that survived = \n',female_survive)
print()
male_survive = len (x[(x['Sex'] == 'male')&(x['Survived'] == 1)])
print('number of men that survived = \n',male_survive)
print()
len_female = len(x[x['Sex'] == 'female'])
len_male = len(x[x['Sex'] == 'male'])
print(len_female)
print()
print(len_male)
percent_survive_female = (female_survive/len_female) * 100
percent_survive_male = (male_survive/len_male) * 100
print()
print('% of female that survived = \n',percent_survive_female)
print()
print('% of male that survived = \n', percent_survive_male)
```

Survival Rate by Passenger Class — Grouped Percentage Calculation

```python
print('x = \n',x.columns)
print()
percent_survive_class = x.groupby('Pclass')['Survived'].mean()*100
print('% of survived from each class : \n',percent_survive_class)
```

Survival Rate by Gender and Passenger Class — Grouped Analysis

```python
print('x = \n',x.columns)
print()
percent_survive_class_gender = x.groupby(['Sex','Pclass'])['Survived'].mean()*100
print('% of survived from each class and gender : \n',percent_survive_class_gender)
```

Average Age Comparison Between Survivors and Non‑Survivors

```python
print('x = \n',x.columns)
print()
a = x.groupby('Survived')['Age'].mean()
a.index=['Dead','Survived']
plt.figure(figsize=(12,6))
a.plot(kind = 'bar',rot=0 , color = ['red','green'])
plt.xlabel('Status')
plt.ylabel('Average Age')
plt.title('Average age of survivors and non survivors')
#plt.savefig("Average_age_of_survivors_and_non_survivors",dpi = 300 , bbox_inches = 'tight' )
plt.show()
```

Survival Rate by Age Groups — Binned Age Analysis

```python
print('x = \n',x.columns)
print()
x['Age']=pd.cut(x['Age'],10)
print(x['Age'])
print()
a = x.groupby('Age')['Survived'].mean()
plt.figure(figsize=(12,6))
a.plot(kind = 'bar',rot=90 , color ='skyblue')
plt.xlabel('Age Group')
plt.ylabel('Survival Rate')
plt.title('Survival Rate by Age Group')
#plt.savefig("Survival_Rate_by_Age_Group",dpi = 300 , bbox_inches = 'tight' )
plt.show()
```

Data Cleaning, Missing Values, Survival Percentages, and Age Outlier Detection

```python
df = pd.read_csv('Titanic-Dataset.csv')   #choose the correct path for yourself
x=df[['Age','Sex','Survived']]
print('x = \n',x.columns)
titanic = x.head(21)
titanic.to_csv('Titanic_sample.csv')    #choose the correct path for yourself
print()
#DESCRIBE + MISSING DATA
describe =x.describe().round(2)
print("describe = \n",describe)
describe.to_csv('describe.csv')   #choose the correct path for yourself
print()
missing = x.isnull().sum()
missing_pct=(missing/len(x))*100
missing_df=pd.DataFrame({'Missing':missing,'Missing %':missing_pct})
missing_df.to_csv('missing.csv')    #choose the correct path for yourself
print(missing_df)
print()
# % OF PPL THAT SURIVED
female_survive = len (x[(x['Sex'] == 'female') & (x['Survived'] == 1)])
print('number of female that survived = \n',female_survive)
print()
male_survive = len (x[(x['Sex'] == 'male')&(x['Survived'] == 1)])
print('number of men that survived = \n',male_survive)
print()
len_female = len(x[x['Sex'] == 'female'])
len_male = len(x[x['Sex'] == 'male'])
print(len_female)
print()
print(len_male)
percent_survive_female = (female_survive/len_female) * 100
percent_survive_male = (male_survive/len_male) * 100
print()
print('% of female that survived = \n',percent_survive_female)
print()
print('% of male that survived = \n', percent_survive_male)
print()

def detect_outliers(a):
    Q1 = a.quantile(0.25)
    Q3 = a.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (a<lower_bound)|(a>upper_bound)
 
age_no_na = df['Age'].dropna()
outliers_age=detect_outliers(age_no_na)

#print the results for outliers

len_age=len(age_no_na)
number_outliers=outliers_age.sum()
pct_outliers=(outliers_age.sum())/len(age_no_na)*100
outliers_value=age_no_na[outliers_age].tolist()
outliers = {
        'Total Values':len_age,
    "Outliers Count":number_outliers,
    'Outliers %':pct_outliers,
    'Outliers':outliers_value

}
outliers_df=pd.DataFrame(outliers)
outliers_df.to_csv('outliers.csv')    #choose the correct path for yourself


print("total of age values: \n",len(age_no_na))   
print("number of outliers in age :\n",outliers_age.sum()) 
print("percentage of outliers in age:\n",(outliers_age.sum())/len(age_no_na)*100)
print("outlier values :\n",age_no_na[outliers_age].tolist())
```
